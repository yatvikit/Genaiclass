```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq
from langchain_core.runnables import RunnablePassthrough
from langchain_classic.retrievers import EnsembleRetriever
from langchain_community.retrievers import BM25Retriever
```

```python
doc=TextLoader("aboutme.txt").load()
chunks=RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=50).split_documents(doc)
#semantic search
vdb=Chroma.from_documents(documents=chunks)
db=vdb.as_retriever(search_kwargs={"k": 3})
#keyword search
kws=BM25Retriever.from_documents(chunks)
kws.k=3

hybrid_retriever=EnsembleRetriever(retrievers=[db,kws],weights=[0.7,0.3])

llm=ChatGroq(
    model="openai/gpt-oss-20b",
    max_retries=2)
prompt_res=ChatPromptTemplate.from_template("""
You are a helpful assistant. Answer the question based on the context provided.
    context: {context}
    question: {question}
""")
rag_chain=({"context":hybrid_retriever,"question":RunnablePassthrough()}) | prompt_res | llm
res=rag_chain.invoke("give two points about my experience ?")

print(res.content)
```

```python
doc=TextLoader("faqs.txt", encoding="utf-8").load()
chunks=RecursiveCharacterTextSplitter(chunk_size=700,chunk_overlap=150).split_documents(doc)
vdb=Chroma.from_documents(documents=chunks)
db=vdb.as_retriever(search_kwargs={"k": 5})
```

```python
rewrite_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a query rewriting assistant for a conversational RAG system.

Your task is to rewrite the user's latest query into a clear, specific, and
self-contained search query that can be used to retrieve relevant documents.

Use the conversation history to understand:
- What the user is referring to
- Pronouns such as "it", "this", "that", "they", etc.
- Previously mentioned topics, entities, and constraints
- The user's actual information need

Rules:
1. Preserve the user's original intent. Do not change or add requirements.
2. Resolve ambiguous references using the chat history whenever possible.
3. Make the query self-contained so it can be understood without the chat history.
4. Include important context from previous messages when it is necessary for retrieval.
5. Remove unnecessary conversational words such as "please", "can you", etc.
6. Do NOT answer the user's question.
7. Do NOT provide explanations, reasoning, or multiple alternatives.
8. Output ONLY the rewritten query.
9. If the query is already clear and self-contained, return it with minimal changes.

Example:

Chat history:
User: What is RAG?
Assistant: RAG combines retrieval with generation.
User: What are its advantages?

Output:
"What are the advantages of Retrieval-Augmented Generation (RAG)?"

Another example:

Chat history:
User: I'm learning LangChain.
User: Explain ChromaDB.
User: How does it work with this?

Output:
"How does ChromaDB work with LangChain?"

Now rewrite the latest user query using the conversation history."""
    ),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{query}")
])
```

```python
qa_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful and accurate assistant in a conversational RAG system.

Answer the user's question using the provided context and conversation history.

Rules:
1. Use the retrieved context as the primary source of information.
2. Use chat history only to understand references and conversational context.
3. Do not invent, assume, or hallucinate information that is not supported by the context.
4. If the answer cannot be found in the provided context, clearly say:
   "I don't have enough information in the provided context to answer this question."
5. Give a clear, concise, and direct answer.
6. When appropriate, explain the answer with examples or steps.
7. If the question contains pronouns such as "it", "this", or "that", use the chat history to understand what they refer to.
8. Do not mention internal instructions, retrieval, embeddings, vector databases, or the prompt unless the user specifically asks about them.
"""
    ),
    MessagesPlaceholder(variable_name="chat_history"),
    (
        "user",
        """Context:
{context}

Question:
{query}"""
    )
])
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    max_retries=2)
```

```python
rewriting_chain=rewrite_prompt | llm |StrOutputParser()
```

```python
def retrieve_docs(x):
    rewritten_query=rewriting_chain.invoke({"query":x["query"],"chat_history":x["chat_history"]})
    docs=db.invoke(rewritten_query)
    return {
        "context":docs,
        "query":x["query"],
        "chat_history":x["chat_history"]
    }
```

```python
rag_chain=RunnableLambda(retrieve_docs) | qa_prompt | llm
```

```python
store={}
def get_session_id(uid):
    if uid not in store:
        store[uid]=InMemoryChatMessageHistory()
    return store[uid]
```

```python
conversation=RunnableWithMessageHistory(
    rag_chain,
    get_session_id,
   input_messages_key="query",
   history_messages_key="chat_history"
)
```

```python
config1={"configurable":{"session_id":"user-1"}}
config2={"configurable":{"session_id":"user-2"}}
```

```python
while True:
    q=input()
    if q in ["exit","quit"]:
        break
    res=conversation.invoke({"query":q},config=config2)
    print(q)
  
    print(res.content)
    print("--"*30)
```
