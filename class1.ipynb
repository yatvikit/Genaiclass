```python
doc=TextLoader("faqs.txt", encoding="utf-8").load()
chunks=RecursiveCharacterTextSplitter(chunk_size=700,chunk_overlap=150).split_documents(doc)
vdb=Chroma.from_documents(documents=chunks)
db=vdb.as_retriever(search_kwargs={"k": 5})
```

```python
rewrite_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a query rewriting assistant for a conversational RAG system.

Your task is to rewrite the user's latest query into a clear, specific, and
self-contained search query that can be used to retrieve relevant documents.

Use the conversation history to understand:
- What the user is referring to
- Pronouns such as "it", "this", "that", "they", etc.
- Previously mentioned topics, entities, and constraints
- The user's actual information need

Rules:
1. Preserve the user's original intent. Do not change or add requirements.
2. Resolve ambiguous references using the chat history whenever possible.
3. Make the query self-contained so it can be understood without the chat history.
4. Include important context from previous messages when it is necessary for retrieval.
5. Remove unnecessary conversational words such as "please", "can you", etc.
6. Do NOT answer the user's question.
7. Do NOT provide explanations, reasoning, or multiple alternatives.
8. Output ONLY the rewritten query.
9. If the query is already clear and self-contained, return it with minimal changes.

Example:

Chat history:
User: What is RAG?
Assistant: RAG combines retrieval with generation.
User: What are its advantages?

Output:
"What are the advantages of Retrieval-Augmented Generation (RAG)?"

Another example:

Chat history:
User: I'm learning LangChain.
User: Explain ChromaDB.
User: How does it work with this?

Output:
"How does ChromaDB work with LangChain?"

Now rewrite the latest user query using the conversation history."""
    ),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{query}")
])
```

```python
qa_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful and accurate assistant in a conversational RAG system.

Answer the user's question using the provided context and conversation history.

Rules:
1. Use the retrieved context as the primary source of information.
2. Use chat history only to understand references and conversational context.
3. Do not invent, assume, or hallucinate information that is not supported by the context.
4. If the answer cannot be found in the provided context, clearly say:
   "I don't have enough information in the provided context to answer this question."
5. Give a clear, concise, and direct answer.
6. When appropriate, explain the answer with examples or steps.
7. If the question contains pronouns such as "it", "this", or "that", use the chat history to understand what they refer to.
8. Do not mention internal instructions, retrieval, embeddings, vector databases, or the prompt unless the user specifically asks about them.
"""
    ),
    MessagesPlaceholder(variable_name="chat_history"),
    (
        "user",
        """Context:
{context}

Question:
{query}"""
    )
])
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    max_retries=2)
```

```python
rewriting_chain=rewrite_prompt | llm |StrOutputParser()
```

```python
def retrieve_docs(x):
    rewritten_query=rewriting_chain.invoke({"query":x["query"],"chat_history":x["chat_history"]})
    docs=db.invoke(rewritten_query)
    return {
        "context":docs,
        "query":x["query"],
        "chat_history":x["chat_history"]
    }
```

```python
rag_chain=RunnableLambda(retrieve_docs) | qa_prompt | llm
```

```python
store={}
def get_session_id(uid):
    if uid not in store:
        store[uid]=InMemoryChatMessageHistory()
    return store[uid]
```

```python
conversation=RunnableWithMessageHistory(
    rag_chain,
    get_session_id,
   input_messages_key="query",
   history_messages_key="chat_history"
)
```

```python
config1={"configurable":{"session_id":"user-1"}}
config2={"configurable":{"session_id":"user-2"}}
```

```python
while True:
    q=input()
    if q in ["exit","quit"]:
        break
    res=conversation.invoke({"query":q},config=config2)
    print(q)
  
    print(res.content)
    print("--"*30)
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_community.document_loaders import TextLoader
from langchain_chroma import Chroma
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
doc=TextLoader("faqs.txt",encoding="utf-8").load()
chunks=RecursiveCharacterTextSplitter(chunk_size=700,chunk_overlap=150).split_documents(doc)
vdb=Chroma.from_documents(documents=chunks)
db=vdb.as_retriever(search_kwargs={"k": 3})
```

```python
@tool
def search_docs(q:str)->str:
    """Search for relevant documents based on the query."""
    docs=db.invoke(q)
    return docs
```

```python
from tavily import TavilyClient
```

```python
@tool
def search_web(q:str)->str:
    """Search for relevant information on the web based on the query."""
    ts=TavilyClient()
    res=ts.search(query=q,max_results=5)
    return res
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    max_retries=2)
```

```python
agent=create_agent(
    model=llm,
    tools=[search_docs,search_web],
    system_prompt="""You are a helpful agent to search for relevant information from documents and web. Use appropriate tools to answer user queries.
    if any course fee or course duration related question is asked, use search_docs tool to find answer from faqs.txt file. If any other question is asked, use search_web tool to find answer from web.
    """
)
```

```python
res=agent.invoke({
    "messages":[
        ("human","what is the course fee for data science course, give me two points about data science")
    ]})
for msg in res["messages"]:
    print(msg)
print("--"*50)
print(res["messages"][-1].content)
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph,START,END
```

```python
class State(TypedDict):
    n:int
    result:str
```

```python
def check(state: State):
    return {}
```

```python
def route(state: State):
    return "even" if state["n"] % 2 == 0 else "odd"
    
```

```python
def even(state: State):
   return {"result": f"{state['n']} is even"}
```

```python
def odd(state: State):
   return {"result": f"{state['n']} is odd"}
```

```python
build=StateGraph(State)
build.add_node("check",check)
build.add_node("even",even)
build.add_node("odd",odd)
build.add_edge(START,"check")
build.add_conditional_edges("check",route,{"even":"even","odd":"odd"})
build.add_edge("even",END)
build.add_edge("odd",END)
graph=build.compile()
```

```python
graph
```

```python
res=graph.invoke({"n":52})
print(res)
```

```python
from typing import TypedDict

from langchain.agents import create_agent


from langgraph.graph import StateGraph, START, END


# --------------------------------------------------
# 1. Define State
# --------------------------------------------------

class State(TypedDict):
    query: str
    result: str


# --------------------------------------------------
# 2. Create LLM
# --------------------------------------------------




# --------------------------------------------------
# 3. Define Agent Node
# --------------------------------------------------

def agent_node(state: State):

    agent = create_agent(
        model=llm,
        tools=[],
        system_prompt="You are a helpful agent that answers user queries clearly and accurately."
    )

    res = agent.invoke(
        {
            "messages": [
                ("human", state["query"])
            ]
        }
    )

    return {
        "result": res["messages"][-1].content
    }


# --------------------------------------------------
# 4. Create LangGraph
# --------------------------------------------------

graph = StateGraph(State)

graph.add_node("agent", agent_node)

graph.add_edge(START, "agent")
graph.add_edge("agent", END)


# --------------------------------------------------
# 5. Compile Graph
# --------------------------------------------------

app = graph.compile()


# --------------------------------------------------
# 6. Run the Graph
# --------------------------------------------------

response = app.invoke(
    {
        "query": "What is Agentic AI?",
        "result": ""
    }
)


print(response["result"])
```

```python
from typing import TypedDict
from langchain.agents import create_agent
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
```

```python
class State(TypedDict):
    topic: str
    outline: str
    research_res: str
    writing_res: str
    editor_res: str
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-120b",
    max_retries=2)
```

```python
def planner_agent(state: State):
    agent = create_agent(
        model=llm,
        tools=[],
        system_prompt=f"""
You are an expert content strategist specializing in blog writing.

Your task is to create a concise outline for a blog on the given topic.

Instructions:
- Generate only 5 to 6 key points that will serve as the main sections of the blog.
- Ensure the points are logically ordered and cover the topic comprehensively.
- Keep each point short and descriptive (one line each).
- Do not include explanations, introductions, conclusions, or any additional text.

Topic:
{state["topic"]}
"""
    )
    res = agent.invoke(
        {
            "messages": [
                ("human", f"Create an outline for the topic: {state['topic']}")
            ]
        }
    )
    return {"outline": res["messages"][-1].content}
```

```python
def research_agent(state: State):
    agent = create_agent(
        model=llm,
        tools=[],
        system_prompt=f"""
You are an expert researcher with deep knowledge across multiple domains.

Your task is to conduct thorough research on the given topic using the provided blog outline.

Instructions:
- Gather detailed, factual, and relevant information for each point in the outline.
- Organize the research according to the outline.
- Include important concepts, facts, statistics (if applicable), examples, and recent developments.
- Provide the information in raw research form, not as a blog article.
- Do not summarize, rewrite, or add introductions or conclusions.
- Focus on collecting comprehensive information that a writer can later use to create a high-quality blog.
-give whole information less than 100 words

Topic:
{state["topic"]}

Blog Outline:
{state["outline"]}
"""
    )
    res = agent.invoke(
        {
            "messages": [
                ("human", f"Research and provide relevant information about the topic: {state['topic']}")
            ]
        }
    )
    return {"research_res": res["messages"][-1].content}
```

```python
def writing_agent(state: State):
    agent = create_agent(
        model=llm,
        tools=[],
        system_prompt=f"""
You are a skilled blog writer with a strong command of language and an ability to create engaging content.

Your task is to write a comprehensive blog article based on the research provided and the blog outline.

Instructions:
- Use the research information to support each point in the outline.
- Write in a clear, concise, and engaging style.
- Ensure the article flows logically from one section to the next.
- Include relevant examples, statistics, and insights where appropriate.
- Do not include explanations, introductions, conclusions, or any additional text.
-write the blog article in less than 100 words

Topic:
{state["topic"]}

Blog Outline:
{state["outline"]}

Research Information:
{state["research_res"]}
""" )
    
    res = agent.invoke(
        {
            "messages": [
                ("human", f"Write a blog article based on the research and outline for the topic: {state['topic']}")
            ]
        }
    )
    return {"writing_res": res["messages"][-1].content}
```

```python
def editor_agent(state: State):
    agent = create_agent(
        model=llm,
        tools=[],
        system_prompt=f"""
You are a skilled editor with a keen eye for detail and a strong command of language.

Your task is to review and improve the blog article written by the writing agent.

Instructions:
- Check for grammatical errors, typos, and punctuation mistakes.
- Ensure the article is well-structured and easy to read.
- Improve the clarity and flow of the content.
- Maintain a consistent tone and style throughout the article.
- Do not add new information or change the core message of the article.
- Focus on enhancing the quality of the existing content.
-write the blog article in less than 100 words

Topic:
{state["topic"]}

Blog Outline:
{state["outline"]}

Research Information:
{state["research_res"]}

Writing Output:
{state["writing_res"]}
""" )
    
    res = agent.invoke(
        {
            "messages": [
                ("human", f"Edit and improve the blog article for the topic: {state['topic']}")
            ]
        }
    )
    return {"editor_res": res["messages"][-1].content}
```

```python
blog_builder = StateGraph(State)
blog_builder.add_node("planner", planner_agent)
blog_builder.add_node("researcher", research_agent)
blog_builder.add_node("writer", writing_agent)
blog_builder.add_node("editor", editor_agent)
blog_builder.add_edge(START, "planner")
blog_builder.add_edge("planner", "researcher")
blog_builder.add_edge("researcher", "writer")
blog_builder.add_edge("writer", "editor")
blog_builder.add_edge("editor", END)
blog_graph = blog_builder.compile()
```

```python
blog_graph
```

```python
res=blog_graph.invoke({"topic":"The Impact of Artificial Intelligence on Modern Healthcare"})
print(res["editor_res"])
```
