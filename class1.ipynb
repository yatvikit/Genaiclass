```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
doc=TextLoader("aboutme.txt").load()
chunks=RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=50).split_documents(doc)
vdb=Chroma.from_documents(documents=chunks)
```

```python
db=vdb.as_retriever(search_kwargs={"k": 3})
```

```python
res=db.invoke("what is price of laptop ?")
print(res)
```

```python
print(len(res))
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt_template = ChatPromptTemplate.from_template("""
You are a helpful assistant. Answer the question based on the context provided.
Context: {context}
Question: {question}
""")
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain=({"context":db,"question":RunnablePassthrough()}) | prompt_template | llm
```

```python
res=rag_chain.invoke("give two points about my experience ?")
print(res.content)
```
